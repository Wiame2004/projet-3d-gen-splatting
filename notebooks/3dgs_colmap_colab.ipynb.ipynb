```python
import glob, cv2

frames = sorted(glob.glob("/content/work/gs_output_3k/train/ours_3000/renders/*.png"))
imgs = [cv2.imread(f) for f in frames]
h, w, _ = imgs[0].shape

out = cv2.VideoWriter("/content/work/demo_3dgs_3k.mp4", cv2.VideoWriter_fourcc(*"mp4v"), 10, (w, h))
# Répéter chaque vue 20 fois (40 frames = 4s à 10 fps)
for im in imgs:
    for _ in range(20):
        out.write(im)
out.release()

from google.colab import files
files.download("/content/work/demo_3dgs_3k.mp4")
```
